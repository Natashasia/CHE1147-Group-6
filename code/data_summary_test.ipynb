```python
# Import packages
from rdkit import Chem
from rdkit.Chem import Draw, Descriptors

import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("raw_data_tox21.csv")
df.head()
```

```python
print("Total rows:", len(df)) # number of molecules
print("Unique molecules:", df["smiles"].nunique()) # check there are no repeated molecules
```

```python
stress_assays = ["SR-ARE", "SR-ATAD5", "SR-HSE", "SR-MMP", "SR-p53"]

print("Molecules tested against >= 1 stress assay:", df[stress_assays].notna().any(axis=1).sum())

print("Molecules teested in all assays:", df[stress_assays].notna().all(axis=1).sum())
```

```python
for assay in stress_assays:
    tested = df[assay].notna().sum() #number of molecules tested in the assay
    positive = int(df[assay].sum())  #number of molecules tested positive in the assay
    print(f"{assay}: tested molecules={tested}, positive result={positive}, positive rate={100*positive/tested:.1f}%")
```
