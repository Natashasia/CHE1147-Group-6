Converting the data set SDF file into a CSV using https://stackoverflow.com/questions/78004477/convert-simple-data-format-file-sdf-to-csv

```python
def sdf_to_csv(sdf_file, output_csv_file, delimiter=','):
    """
    Convert an SDF file to a CSV or TSV file.
    
    Args:
        sdf_file (str): Path to the input SDF file.
        output_file (str): Path to the output CSV or TSV file.
        delimiter (str, optional): Delimiter to use in the output file. Default is ',' (CSV).
    """
    # Read the SDF file
    suppl = Chem.SDMolSupplier(sdf_file)

    # Convert to a pandas DataFrame
    df = PandasTools.LoadSDF(sdf_file, smilesName='SMILES', molColName='Molecule', includeFingerprints=False)

    # Drop the 'Molecule' column if present
    if 'Molecule' in df.columns:
        df = df.drop('Molecule', axis=1)

    # Save the DataFrame as CSV or TSV
    df.to_csv(output_csv_file, sep=delimiter, index=False)
    
    print(f"Conversion complete! Output file: {output_csv_file}")
```

The functions below are declared for adding the mol objects  and cononicalizing isomeric SMILES

```python
def add_mol_object(smile, isomeric:bool=True):
    mol = Chem.MolFromSmiles(smile)
    return mol

def canonicalize_smiles(smiles, isomeric=True):
    if not isinstance(smiles, str) or not smiles.strip():
        return None

    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return None

    return Chem.MolToSmiles(mol, isomericSmiles=isomeric)



df_of_study['mol'] = df_of_study['SMILES'].apply(add_mol_object)
df_of_study['canon_smiles'] = df_of_study['SMILES'].apply(canonicalize_smiles)
```

Next the highly correlated descriptors are addressed, first by updating the correlation matrix with the updated descriptors, and declaring a function to sift through the correlation matrix and remove the highly correlated descriptors

```python
descriptor_columns = df_with_descriptors.columns[10:]

correlation_matrix = df_with_descriptors[descriptor_columns].corr()

def remove_highly_correlated_descriptors(correlation_matrix, threshold=0.90):
    #avoids redundancy by only searching the upper triangle of the correlation matrix to avoid being redundant and avoids the diagonal of self correlation
    top_of_corr_matrix = correlation_matrix.where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
    descriptors_to_remove = [column for column in top_of_corr_matrix.columns if any(top_of_corr_matrix[column].abs() > threshold)]

    reduced_descriptors_matrix = correlation_matrix.drop(descriptors_to_remove, axis=1)

    return reduced_descriptors_matrix

test_reduced_correlation_matrix = remove_highly_correlated_descriptors(correlation_matrix, 0.90)

test_reduced_correlation_matrix
```

**TESTING FOR ANY DESCRIPTORS THAT DO NOT CORRELATE AT ALL WITH THE STRESS RESPONSE INDICATORS**
First samples the columns of the response variables, and then those of the descriptors and checks their correlation with both pearson and spearman methods. Any descriptor that offered less than 10% (0.1) correlation with the stress responses were discarded in an effort to trim down the data set

```python
response_cols = simplified_df_v3.columns[3:8]
descriptor_cols = simplified_df_v3.columns[10:]

numeric_data = simplified_df_v3[list(response_cols) + list(descriptor_cols)].apply(
    pd.to_numeric, errors="coerce"
)
descriptor_cols = [col for col in descriptor_cols if col in numeric_data.columns]

pearson_corr = pd.DataFrame({
    response: numeric_data[descriptor_cols].corrwith(
        numeric_data[response], method="pearson"
    )
    for response in response_cols
})

spearman_corr = pd.DataFrame({
    response: numeric_data[descriptor_cols].corrwith(
        numeric_data[response], method="spearman"
    )
    for response in response_cols
})


correlation_summary = pd.DataFrame({
    "max_abs_pearson": pearson_corr.abs().max(axis=1),
    "max_abs_spearman": spearman_corr.abs().max(axis=1),
})
correlation_summary["max_abs_either"] = correlation_summary.max(axis=1)
correlation_summary = correlation_summary.sort_values(
    "max_abs_either", ascending=False
)

display(correlation_summary.head(20))

threshold = 0.10
candidate_descriptors = correlation_summary.index[
    correlation_summary["max_abs_either"] < threshold
].tolist()

print(f"Candidate descriptors below |correlation| < {threshold}: "
      f"{len(candidate_descriptors)}")
print(candidate_descriptors)
```

**Run only the cells in this new section, top to bottom.** This section is self-contained: it reads `data/tox21_10k_data_all.csv` and computes descriptors for unique canonical structures. You do **not** have to run the teammate's old analysis or the last HTML-minification cell.

**Agreed methodology:**
- **2.1 Labels:** report original records and deduplication; calculate main label/missingness/class-balance table **after canonical-SMILES deduplication**.
- **2.2 Features:** evaluate *all* generated RDKit descriptors for missing/nonfinite values, zero variance and feature–feature correlations. **Identify, but do not remove, zero-variance or highly correlated descriptors.** Calculate full feature statistics; show four chemically interpretable histograms, with 1.5×IQR fences for MolWt, MolLogP and TPSA but not NumHDonors.
- **2.3 Feature–label:** calculate Pearson correlations using nonconstant valid descriptors against each available assay label; show the **top 20 overall** by maximum absolute correlation across the five endpoints. This is **visualization selection, not ML feature selection**.

**Temporary duplicate rule:** Keep the first row for each canonical SMILES, as requested. This can discard complementary/conflicting endpoint labels and should be revisited before final ML evaluation. In this EDA, no other rows or descriptors are dropped solely because of outliers, missing values or pairwise redundancy; unusable values are converted to NaN only for computations.

```python
# SECTION 2.0 — Imports, input checks and isolated file paths.
# Independent of teammate variables: all variables below begin with `s2_`.
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from rdkit import Chem
from rdkit.Chem import Descriptors
from IPython.display import display

s2_input = Path('data/tox21_10k_data_all.csv')
s2_output = Path('eda_section2_outputs')
s2_output.mkdir(parents=True, exist_ok=True)
s2_assays = ['SR-ARE', 'SR-ATAD5', 'SR-HSE', 'SR-MMP', 'SR-p53']

if not s2_input.is_file():
    raise FileNotFoundError(
        f'{s2_input} was not found. Open VS Code/Jupyter in your '
        'CHE1147 project folder (containing the data/ directory).'
    )

s2_raw = pd.read_csv(s2_input)
s2_required = ['SMILES'] + s2_assays
s2_missing_columns = [c for c in s2_required if c not in s2_raw.columns]
if s2_missing_columns:
    raise ValueError(f'Missing expected CSV columns: {s2_missing_columns}')

print(f'Original records: {len(s2_raw):,}')
print(f'Distinct original SMILES: {s2_raw["SMILES"].nunique(dropna=True):,}')
print(f'Available RDKit descriptor functions: {len(Descriptors._descList)}')
print(f'Output folder: {s2_output.resolve()}')
```

## 2.1 Label Analysis (y): canonicalization, deduplication, missing labels, and class balance

Counts based on *original* records document the dataset as supplied. Counts based on the *deduplicated* records are the main results for Sections 2.1–2.3.

```python
# SECTION 2.1A — Convert SMILES to canonical isomeric SMILES and keep FIRST duplicate.
# No attempt to merge repeated labels: this is deliberately provisional.
def s2_parse_canonical(smiles):
    if not isinstance(smiles, str) or not smiles.strip():
        return None
    mol = Chem.MolFromSmiles(smiles)
    return Chem.MolToSmiles(mol, isomericSmiles=True) if mol is not None else None

s2_raw = s2_raw.copy()
s2_raw['s2_canonical_smiles'] = s2_raw['SMILES'].map(s2_parse_canonical)
s2_usable_raw = s2_raw.dropna(subset=['s2_canonical_smiles']).copy()
s2_unique = (
    s2_usable_raw.drop_duplicates(subset='s2_canonical_smiles', keep='first')
    .reset_index(drop=True).copy()
)

s2_raw_n = len(s2_raw)
s2_invalid = s2_raw_n - len(s2_usable_raw)
s2_dup_rows = len(s2_usable_raw) - len(s2_unique)
print(f'Original records: {s2_raw_n:,}')
print(f'Rows with invalid/missing canonical SMILES: {s2_invalid:,}')
print(f'Extra duplicate-structure rows removed (keep first): {s2_dup_rows:,}')
print(f'Unique canonical structures retained: {len(s2_unique):,}')

# Optional diagnostic: count structures with conflicting observed 0/1 labels,
# without changing how duplicates are handled.
s2_conflict_counts = (
    s2_usable_raw.groupby('s2_canonical_smiles')[s2_assays]
    .nunique(dropna=True).gt(1).sum()
)
print('Canonical structures with conflicting nonmissing labels (by assay):')
print(s2_conflict_counts.to_string())

s2_dataset_counts = pd.DataFrame([{
    'original_records': s2_raw_n,
    'distinct_original_smiles': s2_raw['SMILES'].nunique(dropna=True),
    'invalid_or_missing_canonical_smiles_rows': s2_invalid,
    'extra_duplicate_canonical_structure_rows': s2_dup_rows,
    'retained_unique_canonical_structures': len(s2_unique)
}])
s2_dataset_counts.to_csv(s2_output / 'dataset_and_duplicates.csv', index=False)
```

```python
# SECTION 2.1B — Counts and percentages per assay for ORIGINAL and UNIQUE records.
def s2_summarize_labels(frame):
    labels = frame[s2_assays].apply(pd.to_numeric, errors='coerce')
    # Valid assay classes are 0/1; leave all other values as untested/invalid.
    labels = labels.where(labels.isin([0, 1]))
    n_total = len(labels)
    table = pd.DataFrame(index=s2_assays)
    table.index.name = 'Assay'
    table['Total'] = n_total
    table['Labeled'] = labels.notna().sum()
    table['Missing'] = labels.isna().sum()
    table['Missing (%)'] = (100 * table['Missing'] / n_total).round(1)
    table['Inactive'] = labels.eq(0).sum()
    table['Active'] = labels.eq(1).sum()
    table['Active of labeled (%)'] = (
        100 * table['Active'] / table['Labeled'].replace(0, np.nan)
    ).round(1)
    return table

s2_label_original = s2_summarize_labels(s2_raw)
s2_label_unique = s2_summarize_labels(s2_unique)
s2_label_original.to_csv(s2_output / 'table_labels_original_records.csv')
s2_label_unique.to_csv(s2_output / 'table1_labels_deduplicated.csv')
print('Original-record label composition:')
display(s2_label_original)
print('TABLE 1 — Deduplicated label composition (use for proposal):')
display(s2_label_unique)
```

## 2.2 Molecular Descriptor Analysis (X): comprehensive screening, four representative distributions, IQR outliers, descriptor redundancy

The quality, statistics, and pairwise correlations cover **all available RDKit descriptors**. The four plot columns are examples selected for interpretability—not a reduced ML feature set.

```python
# SECTION 2.2A — Compute all RDKit descriptors ONCE per unique canonical structure.
# No reliance on teammate's previous shortened 51-feature dataframe.
s2_descriptor_names = [name for name, fn in Descriptors._descList]

def s2_describe_structure(smiles):
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        return {name: np.nan for name in s2_descriptor_names}
    try:
        return Descriptors.CalcMolDescriptors(mol, missingVal=np.nan, silent=True)
    except Exception:
        return {name: np.nan for name in s2_descriptor_names}

s2_descriptor_records = [
    s2_describe_structure(s) for s in s2_unique['s2_canonical_smiles']
]
s2_X = pd.DataFrame(s2_descriptor_records, columns=s2_descriptor_names)
s2_X = s2_X.apply(pd.to_numeric, errors='coerce')
s2_X = s2_X.replace([np.inf, -np.inf], np.nan)
s2_Y = s2_unique[s2_assays].apply(pd.to_numeric, errors='coerce')
s2_Y = s2_Y.where(s2_Y.isin([0, 1]))

assert len(s2_X) == len(s2_unique) == len(s2_Y)
assert s2_X.index.equals(s2_Y.index)
print(f'Molecules with descriptors: {len(s2_X):,}')
print(f'Descriptor columns: {s2_X.shape[1]:,}')
print(f'Descriptor entries: {s2_X.size:,}')
```

```python
# SECTION 2.2B — Missing/non-finite values, constant features, ALL-descriptor summary stats.
s2_missing_by_feature = s2_X.isna().sum()
s2_nunique = s2_X.nunique(dropna=True)
s2_constant = s2_nunique[s2_nunique == 1].index.tolist()
s2_allmissing = s2_nunique[s2_nunique == 0].index.tolist()
s2_missing_total = int(s2_missing_by_feature.sum())
s2_missing_percent = 100 * s2_missing_total / s2_X.size if s2_X.size else np.nan

s2_quality = pd.DataFrame({
    'missing_or_nonfinite_n': s2_missing_by_feature,
    'missing_or_nonfinite_pct': (100 * s2_missing_by_feature / len(s2_X)).round(3),
    'distinct_nonmissing_values': s2_nunique,
    'constant': s2_nunique.eq(1),
    'all_missing': s2_nunique.eq(0),
})
s2_quality.to_csv(s2_output / 'all_descriptor_quality.csv', index_label='Descriptor')

s2_all_stats = s2_X.describe(percentiles=[.25, .5, .75]).T
s2_all_stats['IQR'] = s2_all_stats['75%'] - s2_all_stats['25%']
s2_all_stats['skewness'] = s2_X.skew()
s2_all_stats.to_csv(s2_output / 'all_descriptor_descriptive_statistics.csv', index_label='Descriptor')

print(f'Descriptors with >=1 missing/non-finite entry: {(s2_missing_by_feature > 0).sum():,}')
print(f'Total missing/non-finite descriptor entries: {s2_missing_total:,} ({s2_missing_percent:.3f}%)')
print(f'Constant descriptors (exactly one observed value): {len(s2_constant)}', s2_constant)
print(f'All-missing descriptors: {len(s2_allmissing)}', s2_allmissing)
print('10 most affected descriptors:')
display(s2_quality.sort_values('missing_or_nonfinite_n', ascending=False).head(10))

# For Pearson correlations only, skip descriptors with no variation.
# We have not deleted these columns from s2_X or s2_all_stats.
s2_usable_features = s2_nunique[s2_nunique > 1].index.tolist()
```

```python
# SECTION 2.2C — ALL-feature Pearson redundancy check (without dropping features).
s2_xx_corr = s2_X[s2_usable_features].corr(method='pearson')
s2_upper = s2_xx_corr.where(
    np.triu(np.ones(s2_xx_corr.shape, dtype=bool), k=1)
)
s2_strong_pairs = (
    s2_upper.stack().loc[lambda r: r.abs() > 0.90]
    .rename('Pearson_r')
    .reset_index()
)
s2_strong_pairs.columns = ['Descriptor_1', 'Descriptor_2', 'Pearson_r']
s2_strong_pairs['abs_r'] = s2_strong_pairs['Pearson_r'].abs()
s2_strong_pairs = s2_strong_pairs.sort_values('abs_r', ascending=False)
s2_strong_pairs.to_csv(s2_output / 'descriptor_pairs_abs_r_gt_0p90.csv', index=False)

print(f'Usable nonconstant descriptors for correlations: {len(s2_usable_features):,}')
print(f'Descriptor pairs with |Pearson r| > 0.90: {len(s2_strong_pairs):,}')
print('10 strongest redundant pairs:')
display(s2_strong_pairs.head(10))
```

```python
# SECTION 2.2D — FIGURE 1: 4 representative histograms; IQR fences for 3 continuous descriptors.
s2_display_features = ['MolWt', 'MolLogP', 'TPSA', 'NumHDonors']
s2_missing_plot_features = [f for f in s2_display_features if f not in s2_X.columns]
if s2_missing_plot_features:
    raise ValueError(f'RDKit descriptors not found: {s2_missing_plot_features}')

s2_outlier_rows = []
for feature in ['MolWt', 'MolLogP', 'TPSA']:
    vals = s2_X[feature].dropna()
    q1, q3 = vals.quantile([0.25, 0.75])
    iqr = q3 - q1
    lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
    count = int(((vals < lower) | (vals > upper)).sum())
    s2_outlier_rows.append({
        'Descriptor': feature,
        'Valid_n': len(vals), 'Q1': q1, 'Q3': q3, 'IQR': iqr,
        'Lower_1p5_IQR': lower, 'Upper_1p5_IQR': upper,
        'Potential_outliers_n': count,
        'Potential_outliers_pct': 100*count/len(vals) if len(vals) else np.nan,
    })
s2_outlier_table = pd.DataFrame(s2_outlier_rows).set_index('Descriptor')
s2_outlier_table.to_csv(s2_output / 'table_1p5IQR_outliers_three_continuous.csv')
print('IQR outlier analysis (unique molecular structures):')
display(s2_outlier_table.round(3))

s2_titles = {
    'MolWt': 'Molecular weight (g/mol)',
    'MolLogP': 'MolLogP',
    'TPSA': 'Topological polar surface area (Å²)',
    'NumHDonors': 'Number of H-bond donors',
}
fig, axes = plt.subplots(2, 2, figsize=(10.5, 6), constrained_layout=True)
for ax, feature in zip(axes.flat, s2_display_features):
    vals = s2_X[feature].dropna()
    if feature == 'NumHDonors':
        counts = vals.astype(int).value_counts().sort_index()
        ax.bar(counts.index, counts.values, width=0.85, color='#477f9f')
        ax.set_xticks(counts.index[::max(1, int(np.ceil(len(counts)/12)))])
    else:
        ax.hist(vals, bins=35, color='#477f9f', edgecolor='white', linewidth=0.4)
        lower = s2_outlier_table.loc[feature, 'Lower_1p5_IQR']
        upper = s2_outlier_table.loc[feature, 'Upper_1p5_IQR']
        # The fences are computed regardless; plot a line only if it falls in the visible range.
        if vals.min() <= lower <= vals.max():
            ax.axvline(lower, color='#c64c47', linestyle='--', linewidth=1.5,
                       label='1.5×IQR fence')
        if vals.min() <= upper <= vals.max():
            ax.axvline(upper, color='#c64c47', linestyle='--', linewidth=1.5,
                       label='1.5×IQR fence')
        n = s2_outlier_table.loc[feature, 'Potential_outliers_n']
        p = s2_outlier_table.loc[feature, 'Potential_outliers_pct']
        ax.text(.98, .96, f'IQR outliers: {n:,} ({p:.1f}%)',
                transform=ax.transAxes, va='top', ha='right', fontsize=8,
                bbox={'facecolor': 'white', 'edgecolor': 'none', 'alpha': .85})
    ax.set_title(feature, fontsize=11, fontweight='bold')
    ax.set_xlabel(s2_titles[feature], fontsize=9)
    ax.set_ylabel('Unique compounds', fontsize=9)
    ax.tick_params(labelsize=8)
    ax.grid(axis='y', alpha=.2)

#fig.suptitle('Figure 1. Molecular descriptor distributions and potential outliers', fontsize=12)
fig.savefig(s2_output / 'figure1_descriptor_histograms.png', dpi=300, bbox_inches='tight')
fig.savefig(s2_output / 'figure1_descriptor_histograms.pdf', bbox_inches='tight')
plt.show()

print('Selected descriptor descriptive statistics:')
display(s2_all_stats.loc[s2_display_features,
    ['count', 'mean', 'std', 'min', '25%', '50%', '75%', 'max', 'IQR', 'skewness']
].round(3))
```

## 2.3 Descriptor–Activity Relationships (X–y): correlations and top-20 heatmap

Only *available* 0/1 labels are used for each endpoint. Pearson correlation with a binary 0/1 label is a point-biserial correlation. The top 20 descriptors shown below are chosen **only for this exploratory figure**; model-feature selection must be fitted on training data within cross-validation later.

```python
# SECTION 2.3 — Pearson correlations with all five endpoints, and FIGURE 2.
from matplotlib.pyplot import bar


s2_xy_corr = pd.DataFrame(index=s2_usable_features)
s2_pair_counts = pd.DataFrame(index=s2_usable_features)
for assay in s2_assays:
    y = s2_Y[assay]
    x = s2_X[s2_usable_features]
    s2_xy_corr[assay] = x.corrwith(y, method='pearson')
    s2_pair_counts[assay] = x.notna().mul(y.notna(), axis=0).sum(axis=0)

s2_xy_corr.to_csv(s2_output / 'all_descriptor_endpoint_pearson_r.csv', index_label='Descriptor')
s2_pair_counts.to_csv(s2_output / 'all_descriptor_endpoint_valid_pair_counts.csv', index_label='Descriptor')

s2_top20_scores = s2_xy_corr.abs().max(axis=1).dropna().nlargest(20)
s2_top20 = s2_xy_corr.loc[s2_top20_scores.index]
s2_top20.to_csv(s2_output / 'figure2_top20_descriptor_endpoint_correlations.csv', index_label='Descriptor')
print('Top-20 descriptors ranked by maximum |Pearson r| across five endpoints:')
display(s2_top20.round(3))
print('Largest observed absolute correlation in displayed set:',
      round(float(s2_top20.abs().max().max()), 3) if not s2_top20.empty else 'none')

if s2_top20.empty:
    print('Cannot plot heatmap: no usable feature–label correlations were available.')
else:
    # Transpose: rows = 5 assays, columns = top 20 descriptors
    heatmap_data = s2_top20.T

    # Wider and shorter figure for the proposal
    fig, ax = plt.subplots(figsize=(15, 5), constrained_layout=True)

    # Symmetric colour scale centered at zero
    limit = max(0.05, float(np.nanmax(np.abs(heatmap_data.to_numpy()))))

    sns.heatmap(
        heatmap_data,
        ax=ax,
        cmap='RdBu_r',
        center=0,
        vmin=-limit,
        vmax=limit,
        linewidths=0.25,
        annot=True,
        fmt='.2f',
        annot_kws={'fontsize': 8},
        cbar_kws={
            'label': 'Pearson r (descriptor vs. activity)',
            'shrink': 0.8
        }
    )
    # Access the colorbar
    cbar = ax.collections[0].colorbar

    # Set colorbar tick label font size
    cbar.ax.tick_params(labelsize=10)

    # Set colorbar title/label font size
    cbar.set_label('Pearson r (descriptor vs. activity)', fontsize=9)

    #ax.set_title(
    #    'Figure 2. Top-20 descriptor–endpoint correlations',
     #   fontsize=12,
     #   pad=10
    #)

    ax.set_xlabel('RDKit molecular descriptor', fontsize=10)
    ax.set_ylabel('Tox21 stress-response assay', fontsize=10)

    # Rotate descriptor labels for readability
    ax.tick_params(axis='x', labelrotation=45, labelsize=8)
    ax.tick_params(axis='y', labelrotation=0, labelsize=9)

    # Align rotated labels
    plt.setp(ax.get_xticklabels(), ha='right', rotation_mode='anchor')

    # Save figure
    fig.savefig(
        s2_output / 'figure2_top20_descriptor_endpoint_heatmap.png',
        dpi=300,
        bbox_inches='tight'
    )

    fig.savefig(
        s2_output / 'figure2_top20_descriptor_endpoint_heatmap.pdf',
        bbox_inches='tight'
    )

    plt.show()
```
